```python
# before running the script, upload the files to the session storage

import pandas as pd
import glob
import os

# Find all .nodes files in the current directory
file_paths = glob.glob('*.nodes')

if not file_paths:
    print("No .nodes files found in the current directory. Please make sure they are uploaded.")
else:
    for file_path in file_paths:
        try:
            # Read the file into a DataFrame using tab separator
            df = pd.read_csv(file_path, sep='\t')

            # Check if the 'Degree' column exists
            if 'Degree' in df.columns:
                avg_degree = df['Degree'].mean()
                print(f"File: {os.path.basename(file_path)} | Average Degree: {avg_degree:.2f}")
            else:
                print(f"File: {os.path.basename(file_path)} | Column 'Degree' not found.")
        except Exception as e:
            print(f"Error processing {file_path}: {e}")
```
