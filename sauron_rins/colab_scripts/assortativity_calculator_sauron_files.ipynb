```python
import pandas as pd
import networkx as nx
import numpy as np
import glob

# Find all .edges files in the current directory
edge_df_names = glob.glob('*.edges')

# Initialize variables required by the code snippet
assortativity_values_tab1 = []

# Loop through all found .edges files
for i, file_name in enumerate(edge_df_names):
    # Load the file into a DataFrame
    df = pd.read_csv(file_name, sep='\t')

    # Construct the graph using the relevant node columns
    G = nx.from_pandas_edgelist(df, source='NodeId1', target='NodeId2')

    # Assortativity calculation block using standard print for warnings
    try:
        assortativity = nx.degree_assortativity_coefficient(G)
        assortativity_values_tab1.append(assortativity)
    except:
        print(f"Warning: Assortativity error for '{file_name}'.")
        assortativity_values_tab1.append(np.nan)

    # Display the resulting calculation for the current file
    print(f"File: {file_name} - Calculated Assortativity: {assortativity_values_tab1[-1]}")
```
